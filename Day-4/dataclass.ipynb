In python the dataclass decorator is a feature introduction in python 3.7 that provides a concise way to define classes primarily intened to store data it automatically generates several special methond such as 

```python
from dataclasses import dataclass
```

```python
@dataclass
class person:
    name:str
    age:int
    profession:str
```

```python
person1 = person("Arun sah", 21, "CSE")
print(person1)
```

```python
@dataclass
class Rectangle:
    length:int
    breath:int
    color:str="white"

l = int(input("Enter the length of rectangle: "))
b = int(input("Enter the breath of rectangle: "))

rectangle1 = Rectangle(l, b)
area = 2 * (rectangle1.length + rectangle1.breath)
print(f"Area of Rectangle is {area} and color is {rectangle1.color}")
print(rectangle1)
```

```python
@dataclass(frozen=True)
class Point:
    x:int
    y:int

point1 = Point(4,6)
print(point1)
point1.x = 8
print(point1)
```

### inheritance in dataclass

```python
@dataclass
class Person:
    name:str
    age:int
@dataclass
class Employee(Person):
    emp_id:str
    depm:str

person2 = Person("kaluwa", 22)
empployee1 = Employee("kaluwa", 22, "123424", "computer science")
empployee1.name
```

### Neasted dataclass

```python
@dataclass
class Address:
    street:str
    city:str
    pincode:str
@dataclass
class Person:
    name:str
    age:int
    address:Address
address = Address("adarsh chock", "malangwa", "88418")
per = Person("Arun sah ", 21, address)
per.address.city
```

```python
@dataclass
class StudentInfo:
    name:str
    roll:int
    enrollment:str
@dataclass
class Marks:
    mid:int
    end:int
    start:int
@dataclass
class Interview(StudentInfo, Marks):
    communication:str
    resume:str
    technical:str
```
