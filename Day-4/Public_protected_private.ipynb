### All the class variables are public 

```python
class Car():
    def __init__ (self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
```

```python
audi = Car(4,5,"Diseal")
```

```python
print(f"The car have {audi.doors} doors and {audi.windows} windows and engine type is {audi.enginetype}")
```

### All the class varaible are protected

```python
class Car1():
    def __init__ (self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
```

```python
class Truck(Car1):
    def __init__ (self, windows, doors, enginetype, horsepower):
        super().__init__(windows, doors, enginetype)
        self.horsepower= horsepower
```

```python
truck = Truck(4,4,"Diesel", 12000)
print(dir(truck))
```

```python
print(truck.enginetype)
```

```python
print(truck.doors)
```

```python
print(truck.windows)
```

```python
print(truck.horsepower)
```
