#### Grading of student using try and except 

```python
class faillAlert(Exception):
    pass

marks = int(input("Enter the marks of student: "))

try:
    if (marks >= 70):
        print("Congurate you have pass the exam with best gread.")
    elif (marks >= 40) & (marks < 70):
        print("Congrate you have been pass with low gread work hard.")
    else:
        raise faillAlert


except faillAlert:
    print("Try again later work hard")
```
